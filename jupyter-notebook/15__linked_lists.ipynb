```python
Linking the nodes:
```

```python
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next = None
        
        
class LinkedList:
    def __init__(self):
        self._start = None                   # head of the list
    
    
    def insert(self, data):
        next_node = ListNode(data)           # new node
        
        
        if self._start is None:              # first time creating a node
            self._current = next_node        # set the start/current to first node
            self._start = next_node
        else:
            self._current.next = next_node   # link the new node
            self._current = next_node        # update current position, start stays same
```

```python
my_list = LinkedList()
```

```python
my_list.insert(25)
```

```python
my_list._start.data
```

```python
my_list.insert(35)
```

```python
my_list._start.next.data
```

```python
# This is getting difficult to work with.Let's simplify this by writing a function that
# pretty prints the list
```

```python
def pretty_print(self):
    out_str = '['
    
    _cur = self._start
    
    while _cur is not None:
        out_str += str(_cur.data) + ", "
        _cur = _cur.next
    return out_str.strip(', ') + ']'
```

```python
LinkedList.pretty_print = pretty_print     # dynamically added a bound method to a class.
```

```python
my_list = LinkedList()
my_list.insert(5)
my_list.insert(7)
my_list.insert(8)
print((my_list).pretty_print())
```

```python
def __str__(self):
    out_str = '['
    
    _cur = self._start
    
    while _cur is not None:
        out_str += str(_cur.data) + ", "
        _cur = _cur.next
    return out_str.strip(', ') + ']'
```

```python
LinkedList.__str__ = __str__     # dynamically added a bound method to a class.
```

```python
my_list = LinkedList()
my_list.insert(5)
my_list.insert(7)
my_list.insert(8)
print(my_list)
```

```python
# we want to call a function on all the nodes.
```

```python
def print_square(node_x):
    print (node_x.data, "x", node_x.data, "=",)
    print (node_x.data * node_x.data)
```

```python
print_square(my_list._start)
```

```python
def call_func(self, fn):
    _cur = self._start
    while _cur is not None:
        fn(_cur)
        _cur = _cur.next
        
LinkedList.call_func = call_func
```

```python
my_list = LinkedList()
my_list.insert(5)
my_list.insert(7)
my_list.insert(8)
print(my_list)

my_list.call_func(print_square)
```
