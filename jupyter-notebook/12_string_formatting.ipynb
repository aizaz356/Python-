```python
STRING FORMATTING:
```

```python
x = 24
y = 97
```

```python
print ("x=", x, "and y=" ,y)
```

```python
ALTERNATIVE METHOD: string formatting
```

```python
print ("x=%d" % x)  # old method (can also add a second variable)
```

```python
print ("x=%d and y=%d" % (x, y))  # old method (can also add a second variable)
```

```python
# Alternatively:
print ("x={} and y={}".format(x, y))     # new method
```

```python
# if you have a float,you need to use a different placeholder for them in the old method.

x = 24.5887676
y = 97
```

```python
print (("x=%f and y=%d") % (x, y)) 
```

```python
print ("x=%.2f and y=%d" % (x,y))
```

```python
print ("x=%.3f and y=%d" % (x,y))
```

```python
Padding:
```

```python
l = [1, 98, 23, 198]
for i in l :
    print("%d" % i)     # can use a padding
```

```python
l = [1, 98, 23, 198]
for i in l :
    print("%4d" % i)     # use a padding/space-padding
```

```python
l = [1, 98, 23, 198]
for i in l :
    print("%7d" % i)     # use a padding/space-padding
```

```python
l = [1, 98, 23, 198]
for i in l :
    print("%04d" % i)   # or can use a 0-padding
```

```python
# Alternatively:
for i in l:
    print ("{:4}".format(i))   # supports padding/space-padding
```

```python
for i in l:
    print ("{:04}".format(i))   # also supports 0-padding
```

```python
s = "Test"
print ("[%s]"%s)           # can also do right-padding with a -
```

```python
s = "Test"
print ("=%s="%s)           
```

```python
s = "Test"
print ("*%s*"%s)           
```

```python
s = "Test"
print ("[%10s]"%s)           # can also do right-padding with a -
```

```python
s = "Test"
print ("[%-10s]"%s)           # can also do right-padding with a -
```

```python
# Alternatively:
# using :n for padding, right padding with a >
print ("={}=".format(s))
```

```python
# using :n for padding, right padding with a >
print ("={:10}=".format(s))
```

```python
# using :n for padding, right padding with a >
print ("={:>10}=".format(s))
```

```python
String formatting into variables:
```

```python
s = "Test"
x = "={:>10}=".format(s)       # can also use the old format here
print(x)
```
